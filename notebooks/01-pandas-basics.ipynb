## Missing values

Many early medalists have no recorded height or weight. pandas stores a missing value as `NaN`, and `isna().sum()` counts them in each column.

```python
df.isna().sum()
```

`dropna()` removes every row that has at least one missing value. Like most pandas methods, it returns a new DataFrame and leaves `df` unchanged.

```python
df2 = df.dropna()
len(df), len(df2)
```

`fillna(value)` replaces every missing value with `value`.

```python
df3 = df.fillna(0)
df3.head()
```

A dictionary fills each column with its own value. pandas matches the keys by column name, so their order does not matter.

```python
df4 = df.fillna({"Age": 21, "Height": 180, "Weight": 70})
df4.head()
```

Filling with a fixed value is useful to learn the method, but in an analysis it invents data: every filled row becomes a 180 cm, 70 kg athlete, and the statistics move toward those values. pandas already skips `NaN` in `mean()`, `median()` and the other statistics, so leaving the values missing, or dropping the rows, is usually the safer choice.

## Select columns and rows

Double brackets select a list of columns.

```python
df5 = df4[["Age", "Height", "Weight"]]
df5.head()
```

`loc` selects by label: a list of row labels, and optionally a list of columns.

```python
df6 = df5.loc[[0, 2, 5]]
df6
```

```python
df7 = df4.loc[[0, 2, 5], ["Age", "Height", "Weight"]]
df7.equals(df6)
```

## Filter rows with a condition

A condition on a column gives one `True` or `False` per row, and indexing with it keeps the `True` rows. Conditions combine with `&` (and) and `|` (or); each one needs its own parentheses.

```python
df8 = df5[(df5["Age"] > 21) & (df5["Age"] < 34)]
df8
```

## Assignment does not copy

`df1 = df` gives the same DataFrame a second name. It does not make a copy, so a change made in place through `df1` would also show up in `df`. `copy()` makes an independent table.

```python
df1 = df
print(df1 is df)

df1 = df.copy()
print(df1 is df)
```
